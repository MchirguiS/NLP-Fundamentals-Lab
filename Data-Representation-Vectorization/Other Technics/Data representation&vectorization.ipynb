```python
import pandas as pd
import numpy as np
```

```python
path = "/content/mixed_reviews_username_review_time.csv"
```

```python
df = pd.read_csv(path, on_bad_lines='skip' )
```

```python
df['review'] = df['review'].str.lower()
```

```python
import re
def remove_html(text):
    return re.sub(r'<.*?>', '', text)
```

```python
df['review'] = df['review'].apply(lambda x: remove_html(x))
```

```python
import re

def remove_urls(text):
    return re.sub(r'(https?://|www\.)\S+', '', text)
```

```python
df['review'] = df['review'].apply(lambda x: remove_urls(x))
```

```python
import string

def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))
```

```python
df['review'] = df['review'].apply(lambda x: remove_punctuation(x))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
df.head()
```

```python
bow = cv.fit_transform(df['review'])
```

```python
print(cv.vocabulary_)
```

**N-Gram**
